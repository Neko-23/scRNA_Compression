```python
import csv
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth

sample = 1
k = 1

in_path = 'compressed/neural/sample'+str(sample)+'/k'+str(k)

with open(in_path + '/high_level_compress/deltas.csv', newline='') as f:
    reader = csv.reader(f)
    dataset = []
    cluster_labels = []
    for i, row in enumerate(reader):
        dataset.append(row[1:])
        cluster_labels.append(row[0])

chosens = []
flags = []
for iter in range(98):
    print('iteration: ', iter)
    te = TransactionEncoder()
    te_ary = te.fit(dataset).transform(dataset)
    df = pd.DataFrame(te_ary, columns=te.columns_)

    s = fpgrowth(df, min_support=0.99-(iter*0.01), use_colnames=True, max_len = 10)

    mlength = 1
    chosen = []
    choice = None
    for ele in s['itemsets']:
        if len(ele) > mlength:
            mlength = len(ele)
            choice = set(ele)
    if choice == None:
        continue
    chosen.append(choice)
    temp_set = choice
    for ele in s['itemsets']:
        ele_set = set(ele)
        if len(ele) > 1 and not temp_set.intersection(ele_set):
            temp_set = temp_set.union(ele_set)
            chosen.append(ele_set)
    chosens += chosen

    deltas = [set(x) for x in dataset]
    flag = []
    for c in chosen:
        f = set()
        for i, d in enumerate(deltas):
            if c.intersection(d) == c:
                deltas[i] = (d - c)
                f.add(i)
        print(f)
        flag.append(f)
    flags += flag

    dataset = [list(x) for x in deltas]

with open(in_path + '/high_level_compress/deltas_further_compressed.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow([len(chosens)]) # number of common item sets
    for chosen in chosens:
        writer.writerow(list(chosen)) # write current commom item set
    for i, d in enumerate(deltas):
        row = [cluster_labels[i]]
        for j, flag in enumerate(flags):
            if i in flag:
                row += ['#'+str(j)]
        row += list(d)
        writer.writerow(row) # 4% reduction 0.01 step 20x 2:06 minutes; 7% reduction 0.01 step 40x 4:55; 23.5% reduction 0.01 step 95x 1:02:00 
```
